## Write a Function

```python
# hacker rank link: "https://www.hackerrank.com/challenges/write-a-function/problem"

def is_leap(year):
    leap = False
    
    if year % 400 == 0:
        leap = True
    elif year % 100 == 0:
        leap = False
    elif year % 4 == 0:
        leap = True
    
    return leap

year = int(input())
print(is_leap(year))
```

## Tuples

```python
# hacker rank link: "https://www.hackerrank.com/challenges/python-tuples/problem"

if __name__ == '__main__':
    n = int(input())
    t = tuple(map(int, input().split()))

    MASK = (1 << 64) - 1
    x = 0x345678
    mult = 1000003
    length = len(t)
    for item in t:
        length -= 1
        x = ((x ^ hash(item)) * mult) & MASK
        mult = (mult + 82520 + length + length) & MASK
    x = (x + 97531) & MASK

    # convert to signed 64-bit
    if x >= 1 << 63:
        x -= 1 << 64
    if x == -1:
        x = -2
    print(x)
```

## Designer Door Mat

```python
# hacker rank link: "https://www.hackerrank.com/challenges/designer-door-mat/problem"

n, m = map(int, input().split())

for i in range(1, n, 2):
    print(('.|.' * i).center(m, '-'))

print('WELCOME'.center(m, '-'))

for i in range(n - 2, 0, -2):
    print(('.|.' * i).center(m, '-'))
```

## String Formatting

```python
# hacker rank link: "https://www.hackerrank.com/challenges/python-string-formatting/problem"

def print_formatted(number):
    width = len(bin(number)[2:])

    for i in range(1, number + 1):
        print(str(i).rjust(width),
              oct(i)[2:].rjust(width),
              hex(i)[2:].upper().rjust(width),
              bin(i)[2:].rjust(width))

if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
```

## Set .discard(), .remove() & .pop()

```python
# hacker rank link: "https://www.hackerrank.com/challenges/py-set-discard-remove-pop/problem"

n = int(input())
s = set(map(int, input().split()))
N = int(input())

for _ in range(N):
    command = input().split()

    if command[0] == "pop":
        s.remove(min(s))
    elif command[0] == "remove":
        s.remove(int(command[1]))
    elif command[0] == "discard":
        s.discard(int(command[1]))

print(sum(s))
```

## Check Strict Superset

```python
# hacker rank link: "https://www.hackerrank.com/challenges/py-check-strict-superset/problem"

A = set(map(int, input().split()))
n = int(input())

result = True

for _ in range(n):
    B = set(map(int, input().split()))

    if not A > B:
        result = False
        break

print(result)
```

## Set Mutations

```python
# hacker rank link: "https://www.hackerrank.com/challenges/py-set-mutations/problem"

n = int(input())
A = set(map(int, input().split()))
N = int(input())

for _ in range(N):
    operation, length = input().split()
    B = set(map(int, input().split()))

    if operation == "intersection_update":
        A.intersection_update(B)
    elif operation == "update":
        A.update(B)
    elif operation == "symmetric_difference_update":
        A.symmetric_difference_update(B)
    elif operation == "difference_update":
        A.difference_update(B)

print(sum(A))
```

## Validating the Phone Number

```python
# hacker rank link: "https://www.hackerrank.com/challenges/validating-the-phone-number/problem?"

import re

n = int(input())

for _ in range(n):
    s = input()
    if re.match(r'^[7-9][0-9]{9}$', s):
        print("YES")
    else:
        print("NO")
```
