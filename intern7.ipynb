## The Captain's Room

```python
# hacker rank link: "https://www.hackerrank.com/challenges/py-the-captains-room/problem"

k = int(input())
rooms = list(map(int, input().split()))

room_set = set(rooms)

captain_room = (sum(room_set) * k - sum(rooms)) // (k - 1)

print(captain_room)
```

## Funny String

```python
# hacker rank link: "https://www.hackerrank.com/challenges/funny-string/problem"

def funnyString(s):
    r = s[::-1]

    for i in range(len(s) - 1):
        if abs(ord(s[i]) - ord(s[i + 1])) != abs(ord(r[i]) - ord(r[i + 1])):
            return "Not Funny"

    return "Funny"


q = int(input())

for _ in range(q):
    s = input()
    print(funnyString(s))
```

## Alternating Characters

```python
# hacker rank link: "https://www.hackerrank.com/challenges/alternating-characters/problem"

def alternatingCharacters(s):
    count = 0

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1

    return count


q = int(input())

for _ in range(q):
    s = input()
    print(alternatingCharacters(s))
```

## Alphabet Rangoli

```python
# hacker rank link: "https://www.hackerrank.com/challenges/alphabet-rangoli/problem" 

def print_rangoli(size):
    import string
    alpha = string.ascii_lowercase
    lines = []

    for i in range(size):
        left = alpha[size - 1:size - i - 1:-1]
        right = alpha[size - i - 1:size]
        row = "-".join(left + right)
        row = row.center(4 * size - 3, "-")
        lines.append(row)

    print("\n".join(lines + lines[-2::-1]))

if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

## Gemstones

```python
# hacker rank link: "http://hackerrank.com/challenges/gem-stones/problem" 

def gemstones(arr):
    common = set(arr[0])

    for rock in arr:
        common = common.intersection(set(rock))

    return len(common)




if __name__ == '__main__':
    n = int(input())
    arr = []

    for _ in range(n):
        arr.append(input())

    print(gemstones(arr))
```

## Beautiful Binary String

```python
# hacker rank link: "https://www.hackerrank.com/challenges/beautiful-binary-string/problem"

def beautifulBinaryString(b):
    count = 0

    while "010" in b:
        b = b.replace("010", "011", 1)
        count += 1

    return count


if __name__ == '__main__':
    n = int(input().strip())
    b = input().strip()

    print(beautifulBinaryString(b))
```

## CamelCase

```python
# hacker rank link: "https://www.hackerrank.com/challenges/camelcase/problem"

def camelcase(s):
    count = 1

    for ch in s:
        if ch.isupper():
            count += 1

    return count


if __name__ == '__main__':
    s = input().strip()
    print(camelcase(s))
```

## Sherlock and the Valid String

```python
# hacker rank link: "https://www.hackerrank.com/challenges/sherlock-and-valid-string/problem"

def isValid(s):
    from collections import Counter

    char_count = Counter(s)
    freq_count = Counter(char_count.values())

    if len(freq_count) == 1:
        return "YES"

    if len(freq_count) != 2:
        return "NO"

    freq1, freq2 = freq_count.keys()
    count1, count2 = freq_count.values()

    if (freq1 == 1 and count1 == 1) or (freq2 == 1 and count2 == 1):
        return "YES"

    if abs(freq1 - freq2) == 1:
        if (freq1 > freq2 and count1 == 1) or (freq2 > freq1 and count2 == 1):
            return "YES"

    return "NO"
if __name__ == '__main__':
    s = input().strip()
    print(isValid(s))
```

## Anagram

```python
# hacker rank link: "https://www.hackerrank.com/challenges/anagram/problem"

def anagram(s):
    if len(s) % 2 != 0:
        return -1

    half = len(s) // 2
    s1 = s[:half]
    s2 = s[half:]

    count = 0

    for ch in s1:
        if ch in s2:
            s2 = s2.replace(ch, "", 1)
        else:
            count += 1

    return count

q = int(input())

for _ in range(q):
    s = input()
    print(anagram(s))
```

## Game of Thrones - I

```python
# hacker rank link: "https://www.hackerrank.com/challenges/game-of-thrones/problem"

def gameOfThrones(s):
    from collections import Counter

    count = Counter(s)
    odd = 0

    for value in count.values():
        if value % 2 != 0:
            odd += 1

    if odd <= 1:
        return "YES"
    else:
        return "NO"

print(gameOfThrones("aaabbbb"))
print(gameOfThrones("cdefghmnopqrstuvw")) 
```

## Caesar Cipher

```python
# hacker rank link: "https://www.hackerrank.com/challenges/caesar-cipher-1/problem"

def caesarCipher(s, k):
    result = ""

    k = k % 26

    for ch in s:
        if ch.islower():
            result += chr((ord(ch) - ord('a') + k) % 26 + ord('a'))
        elif ch.isupper():
            result += chr((ord(ch) - ord('A') + k) % 26 + ord('A'))
        else:
            result += ch

    return result

print(caesarCipher("middle-Outz", 2))
```

## Separate the Numbers

```python
# hacker rank link: "https://www.hackerrank.com/challenges/separate-the-numbers/problem"

def separateNumbers(s):
    if s[0] == '0':
        print("NO")
        return

    for i in range(1, len(s) // 2 + 1):
        first = int(s[:i])
        current = first
        result = ""

        while len(result) < len(s):
            result += str(current)
            current += 1

        if result == s:
            print("YES", first)
            return

    print("NO")

separateNumbers("1234")
separateNumbers("91011")
separateNumbers("99100")
separateNumbers("101103")
```

## Weighted Uniform Strings

```python
# hacker rank link: "https://www.hackerrank.com/challenges/weighted-uniform-string/problem"

def weightedUniformStrings(s, queries):
    weights = set()
    previous = ""
    count = 0

    for ch in s:
        if ch == previous:
            count += 1
        else:
            count = 1
            previous = ch

        weight = (ord(ch) - ord('a') + 1) * count
        weights.add(weight)

    result = []

    for q in queries:
        if q in weights:
            result.append("Yes")
        else:
            result.append("No")

    return result


s = "abccddde"
queries = [1, 3, 12, 5, 9, 10]

print(weightedUniformStrings(s, queries))
```

## Super Reduced String

```python
# hacker rank link: "https://www.hackerrank.com/challenges/reduced-string/problem"

def superReducedString(s):
    result = []

    for ch in s:
        if result and result[-1] == ch:
            result.pop()
        else:
            result.append(ch)

    if not result:
        return "Empty String"

    return "".join(result)


print(superReducedString("aaabccddd"))
print(superReducedString("aa"))
print(superReducedString("baab"))
```
