## Find a String

```python
# hacker rank link: "https://www.hackerrank.com/challenges/find-a-string/problem"

def count_substring(string, sub_string):
    count = 0
    for i in range(len(string) - len(sub_string) + 1):
        if string[i:i + len(sub_string)] == sub_string:
            count += 1
    return count

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()
    
    count = count_substring(string, sub_string)
    print(count)
```

## Introduction to Sets

```python
# hacker rank link: "https://www.hackerrank.com/challenges/py-introduction-to-sets/problem"

def average(array):
    distinct = set(arr)
    return round(sum(distinct) / len(distinct), 3)

if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    result = average(arr)
    print(result)
```

## Nested Lists

```python
# hacker rank link: "https://www.hackerrank.com/challenges/nested-list/problem"

if __name__ == '__main__':
    students = []

    for _ in range(int(input())):
        name = input()
        score = float(input())
        students.append([name, score])

    scores = sorted(set(score for name, score in students))
    second_lowest = scores[1]

    names = sorted(name for name, score in students if score == second_lowest)

    for name in names:
        print(name)
```

## Finding the Percentage

```python
# hacker rank link = "https://www.hackerrank.com/challenges/finding-the-percentage/problem"

if __name__ == '__main__':
    n = int(input())
    student_marks = {}
    for _ in range(n):
        name, *line = input().split()
        scores = list(map(float, line))
        student_marks[name] = scores
    query_name = input()
    average = sum(student_marks[query_name]) / len(student_marks[query_name])
    print(f"{average:.2f}")
```

## Lists

```python
# hacker rank link: "https://www.hackerrank.com/challenges/python-lists/problem"

if __name__ == '__main__':
    N = int(input())
    lst = []

    for _ in range(N):
        command = input().split()

        if command[0] == "insert":
            lst.insert(int(command[1]), int(command[2]))

        elif command[0] == "print":
            print(lst)

        elif command[0] == "remove":
            lst.remove(int(command[1]))

        elif command[0] == "append":
            lst.append(int(command[1]))

        elif command[0] == "sort":
            lst.sort()

        elif command[0] == "pop":
            lst.pop()

        elif command[0] == "reverse":
            lst.reverse()
```

## Collections.Counter()

```python
# hacker rank link: "https://www.hackerrank.com/challenges/collections-counter/problem"

from collections import Counter

X = int(input())
shoes = list(map(int, input().split()))

shoe_count = Counter(shoes)

N = int(input())
money = 0

for i in range(N):
    size, price = map(int, input().split())

    if shoe_count[size] > 0:
        money += price
        shoe_count[size] -= 1

print(money)
```

## The Minion Game

```python
# hacker rank link: "https://www.hackerrank.com/challenges/the-minion-game/problem"

def minion_game(string):
    kevin = 0
    stuart = 0
    vowels = "AEIOU"

    for i in range(len(string)):
        if string[i] in vowels:
            kevin += len(string) - i
        else:
            stuart += len(string) - i

    if kevin > stuart:
        print("Kevin", kevin)
    elif stuart > kevin:
        print("Stuart", stuart)
    else:
        print("Draw")

if __name__ == '__main__':
    s = input()
    minion_game(s)
```

## Merge the Tools!

```python
# hacker rank link: "https://www.hackerrank.com/challenges/merge-the-tools/problem"

def merge_the_tools(string, k):
    for i in range(0, len(string), k):
        sub = string[i:i+k]
        result = ""

        for char in sub:
            if char not in result:
                result += char

        print(result)

if __name__ == '__main__':
    string, k = input(), int(input())
    merge_the_tools(string, k)
```

## Piling Up!

```python
# hacker rank link: "https://www.hackerrank.com/challenges/piling-up/problem"

from collections import deque

for _ in range(int(input())):
    n = int(input())
    blocks = deque(map(int, input().split()))

    top = float('inf')

    while blocks:
        if blocks[0] >= blocks[-1]:
            cube = blocks.popleft()
        else:
            cube = blocks.pop()

        if cube > top:
            print("No")
            break

        top = cube
    else:
        print("Yes")
```
